### 1.	Write a python program to accept two vectors from the user and find their inner product.

```python
def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))

inner_product = dot
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
print('Vector u:', u)
print('Vector v:', v)

print('<u, v>:', inner_product(u, v))
```

### 2.	Write a python program to accept two vectors from the user and check if they are orthogonal.

```python
from math import isclose

def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))

inner_product = dot

def are_orthogonal(vec1, vec2):
    return isclose(inner_product(vec1, vec2), 0, abs_tol=1e-08)
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
print('Vector u:', u)
print('Vector v:', v)

print('Are u & v orthogonal?:', are_orthogonal(u, v))
```

### 3.	Write a python program to accept two vectors from the user and find their norms.

```python
from math import sqrt

def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))

inner_product = dot

def norm(vec):
    return sqrt(inner_product(vec, vec))
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
print('Vector u:', u)
print('Vector v:', v)

print('||u||:', norm(u))
print('||v||:', norm(v))
```

### 4.	Write a python program to accept a vector from the user and normalize it.

```python
from math import sqrt

def scale(scalar, vec):
    return [scalar*elem for elem in vec]

def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))

inner_product = dot

def norm(vec):
    return sqrt(inner_product(vec, vec))

def normalise(vec):
    return scale(1/norm(vec), vec)
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
print('Vector u:', u)
print('Vector v:', v)

print('Normalised u:', normalise(u))
print('Normalised v:', normalise(v))
```

### 5.	Write a python program to accept two vectors from the user and check if they are orthonormal.

```python
from math import isclose, sqrt

def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))

inner_product = dot

def are_orthogonal(vec1, vec2):
    return isclose(inner_product(vec1, vec2), 0, abs_tol=1e-08)

def norm(vec):
    return sqrt(inner_product(vec, vec))

def are_orthonormal(vec1, vec2):
    return are_orthogonal(vec1, vec2) and isclose(norm(vec1), 1, abs_tol=1e-08) and isclose(norm(vec2), 1, abs_tol=1e-08)
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
print('Vector u:', u)
print('Vector v:', v)

print('Are u & v orthonormal?:', are_orthonormal(u, v))
```

### 6.	Write a python program to accept two vectors from the user and find the angle between the two vectors.

```python
from math import sqrt, acos, degrees

def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))

inner_product = dot

def norm(vec):
    return sqrt(inner_product(vec, vec))

def angle(vec1, vec2):
    theta_radians = acos(inner_product(vec1, vec2)/(norm(vec1)*norm(vec2)))
    return degrees(theta_radians)
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
print('Vector u:', u)
print('Vector v:', v)

print('Angle between u & v:', angle(u, v))
```

### 7.	Write a python program to accept a set of vectors from the user and check if the set is orthogonal.

```python
from math import isclose

def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))

inner_product = dot

def are_orthogonal(vec1, vec2):
    return isclose(inner_product(vec1, vec2), 0, abs_tol=1e-08)
```

```python
def is_orthogonal_set(*vecs):
    for i, vec in enumerate(vecs):
        for next in vecs[i+1:]:
            if not are_orthogonal(vec, next):
                return False
    return True
#or
def is_orthogonal_set(*vecs): # alternative one-liner
    return all(are_orthogonal(vec, next) for i, vec in enumerate(vecs) for next in vecs[i+1:])
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
w = input_vec()
print('Vector u:', u)
print('Vector v:', v)
print('Vector w:', w)

print('Is set (u, v, w) orthogonal?:', is_orthogonal_set(u, v, w))
```
