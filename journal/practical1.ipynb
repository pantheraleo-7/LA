# Operations on vectors

### 1.	Write a python program to accept two vectors from the user and add them. 

```python
def add(*vecs):
    return [sum(elems) for elems in zip(*vecs, strict=True)]
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
print('Vector u:', u)
print('Vector v:', v)

print('u+v:', add(u, v))
```

### 11.	Write a MENU DRIVEN python program to accept two vectors from the user and perform the following operations: Add, Subtract, scale & dot product.

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

def scale(scalar, vec):
    return [scalar*elem for elem in vec]

def add(*vecs):
    return [sum(elems) for elems in zip(*vecs, strict=True)]

def sub(vec1, vec2):
    return [elem1-elem2 for elem1, elem2 in zip(vec1, vec2, strict=True)]

def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))
```

```python
while True:
    inp = input('What operation would you like to perform?\
                \n0-exit \n1-addition \n2-subtraction \n3-scalar multiplication \n4-dot product\
                \nEnter: ')
    
    if inp=='0':
        break
    elif inp=='1':
        u, v = input_vec(), input_vec()
        print('Vector u:', u)
        print('Vector v:', v)
        print('u+v:', add(u, v))
    elif inp=='2':
        u, v = input_vec(), input_vec()
        print('Vector u:', u)
        print('Vector v:', v)
        print('u-v:', sub(u, v))
    elif inp=='3':
        s, u = int(input("Enter the scalar multiplier: ")), input_vec()
        print('Vector u:', u)
        print(f"{s}u: ", scale(s, u))
    elif inp=='4':
        u, v = input_vec(), input_vec()
        print('Vector u:', u)
        print('Vector v:', v)
        print('u.v:', dot(u, v))
    else:
        print('Enter a valid input')
```

### 12.	Write a python program to accept two vectors from the user and generate a linear combination of the two.

```python
def scale(scalar, vec):
    return [scalar*elem for elem in vec]

def add(*vecs):
    return [sum(elems) for elems in zip(*vecs, strict=True)]
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
print('Vector u:', u)
print('Vector v:', v)

scaled_u = scale(2, u)
scaled_v = scale(4, v)
print('2u:', scaled_u)
print('4v:', scaled_v)

print('linear combination of u & v = 2u+4v:', add(scaled_u, scaled_v))
```
