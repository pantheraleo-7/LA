### 6.	Write a python program to accept an R x C matrix from the user and print/display the specified row/column.

```python
def transpose(mat):
    r, c = len(mat), len(mat[0])
    
    ret_mat = []
    for j in range(c):
        ret_row = [row[j] for row in mat]
        ret_mat.append(ret_row)
    
    return ret_mat

def get_row(mat, row):
    return mat[row-1]

def get_col(mat, col):
    return transpose(mat)[col-1]
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = input_mat()
print('Matrix A:', A)
row = int(input('Enter row: '))
col = int(input('Enter column: '))

print(f'row {row} of Matrix A:', get_row(A, row))
print(f'column {col} of Matrix A:', get_col(A, col))
```

### 7.	Write a python program to accept an R x C matrix and a scalar from the user. Find the scalar matrix multiplication.

```python
def scale(scalar, mat):
    ret_mat = []
    for row in mat:
        ret_row = [scalar*elem for elem in row]
        ret_mat.append(ret_row)
    
    return ret_mat
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = input_mat()
print('Matrix A:', A)
s = int(input('Input scalar: '))

print(f'{s}A:', scale(s, A))
```

### 8.	Write a python program to accept two R x C matrices from the user. Find the sum of the two matrices.

```python
def add(*mats):
    ret_mat = []
    for rows in zip(*mats, strict=True):
        ret_row = [sum(elems) for elems in zip(*rows, strict=True)]
        ret_mat.append(ret_row)
    
    return ret_mat
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = input_mat()
B = input_mat()
print('Matrix A:', A)
print('Matrix B:', B)

print('A+B:', add(A, B))
```

### 9.	Write a python program to accept two R x C matrices ‘A’ and ‘B’ from the user. Find the matrix 2A + B.

```python
def scale(scalar, mat):
    ret_mat = []
    for row in mat:
        ret_row = [scalar*elem for elem in row]
        ret_mat.append(ret_row)
    
    return ret_mat

def add(*mats):
    ret_mat = []
    for rows in zip(*mats, strict=True):
        ret_row = [sum(elems) for elems in zip(*rows, strict=True)]
        ret_mat.append(ret_row)
    
    return ret_mat
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = input_mat()
B = input_mat()
print('Matrix A:', A)
print('Matrix B:', B)

scaled_A = scale(2, A)
print('2A:', scaled_A)
print('2A+B:', add(scaled_A, B))
```

### 10.	Write a python program to accept two matrices A, B from the user, check if AB is well-defined and hence find AB.

```python
def mul(mat1, mat2):
    mat2_T = transpose(mat2)
    ret_mat = []
    for row in mat1:
        ret_row = []
        for col in mat2_T:
            ret_row.append(sum(elem1*elem2 for elem1, elem2 in zip(row, col, strict=True)))
        ret_mat.append(ret_row)
    
    return ret_mat
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = input_mat()
B = input_mat()
print('Matrix A:', A)
print('Matrix B:', B)

print('AB:', mul(A, B))
```

### 11.	Write a python program to accept three 3 x 3 matrices ‘A’, ‘B’ and ‘C’ from the user. Find the matrix A + B + C.

```python
def add(*mats):
    ret_mat = []
    for rows in zip(*mats, strict=True):
        ret_row = [sum(elems) for elems in zip(*rows, strict=True)]
        ret_mat.append(ret_row)
    
    return ret_mat
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = input_mat()
B = input_mat()
C = input_mat()
print('Matrix A:', A)
print('Matrix B:', B)
print('Matrix C:', C)

print('A+B+C:', add(A, B, C))
```
