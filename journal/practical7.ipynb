### 1.	Write a python program to find all the eigenvalues of a matrix.

```python
from sympy import Symbol, eye, det, roots, Matrix, S

def get_A_LI(mat, var='lamda'):
    r, c = mat.shape
    L = Symbol(var)
    I = eye(r)
    
    return mat - L*I

def charac_poly(mat, var='lamda'):
    A_LI = get_A_LI(mat, var=var)
    
    return det(A_LI).as_poly().monic()

def eig_vals(mat):
    charac = charac_poly(mat)
    
    return roots(charac)
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = Matrix(input_mat(square=True, dtype=S))
print('Matrix A:', A)

print('Eigenvalues of A with their algebraic multiplicity:', eig_vals(A))
```

### 2.	Write a python program to accept two 2 x 2 matrices ‘A’ and ‘B’ from the user. Find the eigenvalues of the matrix A + 2B.

```python
def scale(scalar, mat):
    ret_mat = []
    for row in mat:
        ret_row = [scalar*elem for elem in row]
        ret_mat.append(ret_row)
    
    return ret_mat

def add(*mats):
    ret_mat = []
    for rows in zip(*mats):
        ret_row = [sum(elems) for elems in zip(*rows, strict=True)]
        ret_mat.append(ret_row)
    
    return ret_mat

from sympy import Symbol, eye, det, roots, Matrix, S

def get_A_LI(mat, var='lamda'):
    r, c = mat.shape
    L = Symbol(var)
    I = eye(r)
    
    return mat - L*I

def charac_poly(mat, var='lamda'):
    A_LI = get_A_LI(mat, var=var)
    
    return det(A_LI).as_poly().monic()

def eig_vals(mat):
    charac = charac_poly(mat)
    
    return roots(charac)
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = input_mat(square=True)
B = input_mat(square=True)
print('Matrix A:', A)
print('Matrix B:', B)

scaled_B = scale(2, B)
print('2B:', scaled_B)

sum_mats = add(A, scaled_B)
print('A+2B:', sum_mats)

mat = Matrix(sum_mats, dtype=S)

print('Eigenvalues of A+2B with their algebraic multiplicity:', eig_vals(mat))
```

### 3.	Write a python program to find the eigenvalues & corresponding eigenvectors of a matrix.

```python
from sympy import Symbol, eye, det, roots, Matrix, S

def get_A_LI(mat, var='lamda'):
    r, c = mat.shape
    L = Symbol(var)
    I = eye(r)
    
    return mat - L*I

def eigs(mat):
    A_LI = get_A_LI(mat)
    charac = det(A_LI)
    vals = roots(charac)
    vecs = [A_LI.subs('lamda', val).nullspace() for val in vals]
    vals = {val: (alg_mcity, len(spanset))
            for (val, alg_mcity), spanset in zip(vals.items(), vecs)}
    
    return vals, vecs
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = Matrix(input_mat(square=True, dtype=S))
print('Matrix A:', A)

print('Eigenvalues of A with their algebraic & geometric multiplicity and eigenvectors respectively:\n', eigs(A))
```

### 6.	Write a python program to find its diagonal matrix and modal matrix.

```python
import numpy as np
from sympy import Symbol, eye, det, roots, Matrix, S

def get_A_LI(mat, var='lamda'):
    r, c = mat.shape
    L = Symbol(var)
    I = eye(r)
    
    return mat - L*I

def eigs(mat):
    A_LI = get_A_LI(mat)
    charac = det(A_LI)
    vals = roots(charac)
    vecs = [A_LI.subs('lamda', val).nullspace() for val in vals]
    vals = {val: (alg_mcity, len(spanset))
            for (val, alg_mcity), spanset in zip(vals.items(), vecs)}
    
    return vals, vecs

def diagonalise(mat):
    vals, vecs = eigs(mat)
    val_list = [val for val in vals for _ in range(vals[val][0])]
    vec_list = [vec for spanset in vecs for vec in spanset]
    
    if len(vec_list)<len(val_list):
        return False
    return np.diag(val_list), np.column_stack(vec_list)
```

```python
def input_mat(square=False, dtype=float):
    if square:
        r = c = int(input('Enter the dimension of the square matrix: '))
    else:
        r = int(input('Enter the number of rows: '))
        c = int(input('Enter the number of columns: '))
    
    ret_mat = []
    for i in range(r):
        ret_row = [dtype(input(f'Enter element for position [{i+1}, {j+1}]: ')) for j in range(c)]
        ret_mat.append(ret_row)
    
    return ret_mat

A = Matrix(input_mat(square=True, dtype=S))
print('Matrix A:', A)

print('Diagonal & Modal Matrices of A:\n', diagonalise(A))
```
