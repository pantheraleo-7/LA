# Gram Schmidt Orthogonalization process

### 1.	Write a python program to implement Gram Schmidt Orthogonalization process on a set of vectors.

```python
from math import sqrt

def scale(scalar, vec):
    return [scalar*elem for elem in vec]

def sub(vec1, vec2):
    return [elem1-elem2 for elem1, elem2 in zip(vec1, vec2, strict=True)]

def dot(vec1, vec2):
    return sum(elem1*elem2 for elem1, elem2 in zip(vec1, vec2, strict=True))

inner_product = dot

def norm(vec):
    return sqrt(inner_product(vec, vec))

def gram_schmidt(*vecs):
    ortho_set = []
    for vec in vecs:
        res = vec
        for ortho_vec in ortho_set:
            proj = scale(inner_product(vec, ortho_vec)/norm(ortho_vec)**2, ortho_vec)
            res = sub(res, proj)
        ortho_set.append([round(elem, 8) for elem in res])
    
    return ortho_set
```

```python
def input_vec(dtype=float):
    dim = int(input('Enter dimension of vector: '))
    return [dtype(input(f'Enter element {i+1}: ')) for i in range(dim)]

u = input_vec()
v = input_vec()
w = input_vec()
print('Set (u, v, w):', (u, v, w))

print('Orthogonalised set from set (u, v, w):', gram_schmidt(u, v, w))
```
